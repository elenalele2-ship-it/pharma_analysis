```python
import pandas as pd
```

```python
sales_test = pd.read_csv('sales_test_2025.csv')
```

```python
sales_test.head()
sales_test.dtypes
```

```python
sales_test.head()
```

```python
#selezioniamo solo le righe sales

sales_only25 = sales_test [sales_test ['Type']== 'Sales']
```

```python
sales_only25.head()
```

```python
#facciamo il melt e trasformiamo wide in long

sales_train_2025_long = sales_only25.melt (
id_vars = ['ProductID'],
value_vars = sales_only25.columns[3:],
var_name ='date' ,
value_name = 'quantity_sales_2025'
)

sales_train_2025_long.head()
```

```python
# date problem

sales_train_2025_long ['date'] = pd.to_datetime(sales_train_2025_long  ['date'])
```

```python
sales_train_2025_long.rename(columns={'quantity_sales_2025': 'quantity_sales'}, inplace=True)
```

```python
print(sales_train_2025_long.dtypes)
print(sales_train_2025_long.head())
```

```python
sales_train_2025_long.to_csv('sales_train_2025_long.csv', index=False)
```

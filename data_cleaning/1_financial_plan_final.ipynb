```python
import pandas as pd
```

```python
financial_plan = financial_plan.drop(columns=['Product_name', 'Price_PLN_per_unit'])
```

```python
financial_plan.head()
```

melt and Pivot (23/24) e 2025


```python
# Select specific columns
selected_columns_2324 = financial_plan[
    ['ProductID',
     'Plan_2023_units',
     'Plan_2024_units',
     'Plan_2023_PLN',
     'Plan_2024_PLN'
     ]]
display(selected_columns_2324.head())
```

```python
# Select specific columns
selected_columns_2025 = financial_plan[
    ['ProductID',
    'Plan_2025_units',
     'Plan_2025_PLN'
     ]]
display(selected_columns_2025.head())
```

```python
# Melt the 2023 and 2024 data
melted_2324 = selected_columns_2324.melt(
    id_vars=['ProductID'],
    value_vars=['Plan_2023_units', 'Plan_2024_units', 'Plan_2023_PLN', 'Plan_2024_PLN'],
    var_name='Variable',
    value_name='Value'
)
display(melted_2324.head())
```

```python
# Melt the 2025 data
melted_2025 = selected_columns_2025.melt(
    id_vars=['ProductID'],
    value_vars=['Plan_2025_units', 'Plan_2025_PLN'],
    var_name='Variable',
    value_name='Value'
)
display(melted_2025.head())
```

```python


melted_2324[['Prefix', 'Year', 'Measure']] = melted_2324['Variable'].str.split('_', expand=True)


display(melted_2324.head())
```

```python
melted_2324['Year'] = melted_2324['Year'].astype(int)
melted_2324 = melted_2324.drop(columns=['Variable', 'Prefix'])
display(melted_2324.head())
```

```python
melted_2025[['Prefix', 'Year', 'Measure']] = melted_2025['Variable'].str.split('_', expand=True)
melted_2025['Year'] = melted_2025['Year'].astype(int)
melted_2025 = melted_2025.drop(columns=['Variable', 'Prefix'])
display(melted_2025.head())
```

```python
melted_2324.dtypes
melted_2025.dtypes
```

```python
financial_plan_2324_final = melted_2324.pivot(
    index=['ProductID', 'Year'], columns='Measure', values='Value')
```

```python
melted_2324.head()
```

```python
financial_plan_2324_final.head()
```

```python
financial_plan_2324_final = financial_plan_2324_final.reset_index()
display(financial_plan_2324_final.head())
financial_plan_2324_final.dtypes
```

```python
financial_plan_2025_final = melted_2025.pivot(
    index=['ProductID', 'Year'], columns='Measure', values='Value')
display(financial_plan_2025_final.head())
```

```python
financial_plan_2025_final = financial_plan_2025_final.reset_index()
display(financial_plan_2025_final.head())
financial_plan_2025_final.dtypes
```

```python
financial_plan_2025_final = financial_plan_2025_final.reset_index()
display(financial_plan_2025_final.head())
financial_plan_2025_final.dtypes
```

```python
financial_plan_2324_final.isnull().sum()
financial_plan_2025_final.isnull().sum()
```

```python
financial_plan_2324_final.shape
```

```python
financial_plan_2025_final.shape
```

```python
financial_plan_2324_final['units'] = financial_plan_2324_final['units'].astype(int)
financial_plan_2025_final['units'] = financial_plan_2025_final['units'].astype(int)
```

```python
financial_plan_2324_final.to_csv('financial_plan_2324_final.csv', index=False)
financial_plan_2025_final_cleaned = financial_plan_2025_final.drop(columns=['index'])
financial_plan_2025_final_cleaned.to_csv('financial_plan_2025_final.csv', index=False)
print('Files saved successfully!')
```
